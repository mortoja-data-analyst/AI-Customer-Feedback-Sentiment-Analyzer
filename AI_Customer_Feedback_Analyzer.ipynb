```python
# ===================================================
# PROJECT 3: AI CUSTOMER FEEDBACK & SENTIMENT ANALYZER
# AUTHOR: MORTOJA SARDAR (DATA ANALYST)
# ARTIFICIAL INTELLIGENCE & TEXT ANALYTICS
# ===================================================

# STEP 1: INSTALLING NECESSARY AI TOOLS (DO NOT REMOVE THIS LINE)
!pip install gTTS -q

import time
import IPython.display as ipd
from gtts import gTTS

print("=" * 60)
print("     🌟 AI CUSTOMER FEEDBACK & SENTIMENT ANALYZER 🌟     ")
print("=" * 60 + "\n")

# STEP 2: INPUT CUSTOMER REVIEW DATA
customer_review = "I am highly satisfied with this product! The quality is amazing and delivery was super fast."

print(f"📥 [INPUT FEEDBACK] : \"{customer_review}\"")
print("📊 [SYSTEM STATUS]  : Analyzing text patterns and emotions...")
time.sleep(2)

# STEP 3: CORE AI SENTIMENT DETECTION LOGIC
review_lower = customer_review.lower()

if "amazing" in review_lower or "satisfied" in review_lower or "good" in review_lower or "great" in review_lower:
    sentiment_result = "POSITIVE (Happy Customer) 😊"
    voice_alert = "Success! The customer feedback is highly positive. Great job team!"
elif "bad" in review_lower or "broken" in review_lower or "worst" in review_lower or "disappointed" in review_lower:
    sentiment_result = "NEGATIVE (Unhappy Customer) 😡"
    voice_alert = "Alert! The customer feedback is negative. Please check the issue immediately."
else:
    sentiment_result = "NEUTRAL (Average Feedback) 😐"
    voice_alert = "The customer feedback is neutral."

# PRINTING THE FINAL ANALYTICS REPORT
print("-" * 60)
print(f"🔍 ANALYSIS TARGET   :  CUSTOMER SENTIMENT")
print("-" * 60)
print(f"📈 DETECTED SENTIMENT:  {sentiment_result}")
print("-" * 60 + "\n")

# STEP 4: AI VOICE REPORT SYNTHESIS
print("🔊 [AI ENGINE] Synthesizing Voice Analytics Report...")
tts = gTTS(voice_alert, lang='en')
tts.save("customer_sentiment_report.mp3")

print("✨ [SUCCESS] AI Sentiment Report and Audio Output are Ready:")
print("=" * 60)
ipd.Audio("customer_sentiment_report.mp3")
```
